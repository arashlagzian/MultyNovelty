```python
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
embedding_model = SentenceTransformer('all-MiniLM-L6-v2', device=device)  

model_name = "gpt2-medium"
tokenizer = AutoTokenizer.from_pretrained(model_name)
model = AutoModelForCausalLM.from_pretrained(model_name).to(device)

generator = pipeline(
    "text-generation",
    model=model,
    tokenizer=tokenizer,
    device=0 if torch.cuda.is_available() else -1  
)
```

```python
num_samples = 100     # 100, 250, 500, 1000, 1500, 2000
max_length = 125


TEMPERATURE = 0.9
TOP_P = 0.95

CHUNK_SIZE = 10
```

```python

def remove_prompt_from_output(prompt_text, output_text):
    """
    If the output starts with the prompt text, remove it. Otherwise return as-is.
    """
    if output_text.startswith(prompt_text):
        return output_text[len(prompt_text):].strip()
    return output_text.strip()
```

```python
_global_batched_responses = None

def generate_outputs(prompt, num_samples=num_samples):
    global _global_batched_responses

    if _global_batched_responses is not None:
        return _global_batched_responses[prompt]

    _global_batched_responses = {p: [None]*num_samples for p in prompts}

    print(f"\n[INFO] BATCH MODE: Generating {num_samples} outputs for each of {len(prompts)} prompts.\n")

    for sample_idx in range(num_samples):
        print(f"\n=== Chunk / Iteration {sample_idx+1} of {num_samples} ===")

        expanded_prompts = []
        debug_views_for_this_chunk = {}

        for p in prompts:
            possible_views = text_views.get(p, [])
            chosen_view = None
            if possible_views:
                chosen_view = random.choice(possible_views)
                expanded_prompt = f"{p}\n\nVIEW: {chosen_view}\n"
            else:
                expanded_prompt = p

            expanded_prompts.append(expanded_prompt)
            debug_views_for_this_chunk[p] = chosen_view

        for i, ep in enumerate(expanded_prompts):
            ep_preview = ep[:].replace('\n', ' ')
           

        raw_batch_res = generator(
            expanded_prompts,
            max_length=max_length,
            temperature=TEMPERATURE,
            top_p=TOP_P,
            do_sample=True,
            pad_token_id=tokenizer.eos_token_id,
            truncation=True,
            num_return_sequences=1
        )


        for i, sublist in enumerate(raw_batch_res):
            p = prompts[i]
            out_dict = sublist[0]
            gen_text = out_dict["generated_text"]

            gen_text_preview = gen_text[:].replace('\n', ' ')


            gen_text = remove_prompt_from_output(expanded_prompts[i], gen_text)
            
            _global_batched_responses[p][sample_idx] = gen_text


    print("\n[BATCH MODE] Done generating all prompts.\n")
    return _global_batched_responses[prompt]
```

```python
from IPython.core.interactiveshell import InteractiveShell
InteractiveShell.ast_node_interactivity = "all"
```
