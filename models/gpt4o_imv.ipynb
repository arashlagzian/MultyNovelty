```python

API_KEY = ""  
API_ENDPOINT = ""
HEADERS = {
    "Authorization": f"Bearer {API_KEY}",
    "Content-Type": "application/json",
    "HTTP-Referer": "<YOUR_SITE_URL>",  # Optional
    "X-Title": "<YOUR_SITE_NAME>",  # Optional
}

def generate_outputs_api(model_name, prompt, num_samples, max_length, catch, max_retries=20):
    outputs = []
    generated_count = 0

    while generated_count < num_samples:
        samples_to_generate = min(catch, num_samples - generated_count)
        retries = 0
        wait_time = 2  

        while retries < max_retries:
            data = {
                "model": model_name,
                "messages": [
                    {"role": "system", "content": "You are a helpful assistant.Your answers should be plain text without using special characters, bullet points, or lists."}, # Don't repeat the question in your answer and just provide the answer at most 125 tokens.
                    {"role": "user", "content": prompt},
                ],
                "max_tokens": max_length,
                "temperature": 0.9,
                "top_p": 0.95,
                "n": samples_to_generate,
            }

            try:
                response = requests.post(API_ENDPOINT, json=data, headers=HEADERS, timeout=30)
                response.raise_for_status()
                response_data = response.json()

                temp_outputs = [
                    choice.get("message", {}).get("content", "").strip()
                    for choice in response_data.get("choices", [])
                ]

                if all(temp_outputs) and len(temp_outputs) == samples_to_generate:
                    outputs.extend(temp_outputs)
                    generated_count += len(temp_outputs)

                    break  
                else:
           
                    print(f"Invalid response received, retrying in {wait_time}s... ({retries+1}/{max_retries})")
                    retries += 1
                    time.sleep(wait_time)
                    wait_time = 2 

            except requests.exceptions.Timeout as e:
            
                print(f"Request timed out: {e}. Retrying in {wait_time}s... ({retries+1}/{max_retries})")
                retries += 1
                time.sleep(wait_time)
                wait_time = 2

            except requests.exceptions.RequestException as e:
       
                print(f"API request error: {e}, retrying in {wait_time}s... ({retries+1}/{max_retries})")
                retries += 1
                time.sleep(wait_time)
                wait_time = 2

        if retries == max_retries:
            print(f"Failed to generate a valid response after {max_retries} attempts. Skipping this batch.")

    return outputs



model_names = ["gpt-4o"]
num_samples = [2000] 
max_length = 125

global_generated_count = 0

for model_name in model_names:
    for sample_size in num_samples:
        output_file = f"gpt4o_imv_{sample_size}{max_length}.json"


        if os.path.exists(output_file):
            with open(output_file, "r") as f:
                all_outputs = json.load(f)
            print(f"Loaded existing results from {output_file}.")
        else:
            all_outputs = {}

   
        for prompt in prompts:
           
            existing_count = len(all_outputs.get(prompt, []))
            
            
            if existing_count >= sample_size:
                print(f"Skipping '{prompt}' (already have {existing_count}/{sample_size} samples).")
                continue
           
            needed = sample_size - existing_count
            print(f"Prompt: '{prompt}' has {existing_count} samples; need {needed} more ...")

            new_responses = []

            for _ in range(needed):
                views_for_prompt = image_views.get(prompt, [])
                if views_for_prompt:
                    selected_view = random.choice(views_for_prompt)

                    combined_prompt = (
                        f"{prompt}\n"
                        "Your answers should be plain text and use details in this image description in your answer:\n"
                        f"{selected_view}"
                    )
                else:
                    combined_prompt = prompt
   
                single_output = generate_outputs_api(
                    model_name=model_name,
                    prompt=combined_prompt,
                    num_samples=1,
                    max_length=max_length,
                    catch=1
                )
           
                if single_output:
                    
                    new_responses.append(single_output[0])
                    global_generated_count += 1

                    if global_generated_count % 10 == 0:
                        print(f"Global progress: {global_generated_count} outputs generated so far...")

            if prompt not in all_outputs:
                all_outputs[prompt] = []
            all_outputs[prompt].extend(new_responses)
           
            print(f"Prompt: '{prompt}' | Total responses generated so far: {len(all_outputs[prompt])}")
         
            with open(output_file, "w") as f:
                json.dump(all_outputs, f, indent=4)

            print(f"Partial results saved to {output_file} for prompt '{prompt}'\n")

        print(f"Final outputs saved to {output_file}")
```
